```python
# Order the file names
sorted_id1 = sorted(mappingDicom.keys())
sorted_id2 = sorted(mappingHeader.keys())
```

```python
# Go through each file name, check if same
# Compare Patient id from file mapped from dicom mapping with header mapped from header mapping
match = 0
for i, (id1, id2) in enumerate(zip(sorted_id1, sorted_id2)):
    if id1 != id2 :
        print(f"Mismatch at index {i}: {id1} != {id2}")
    else :
        path = Path(mappingDicom[id1])
        ds = dcmread(path)
        if ds.PatientID != mappingHeader[id2].PatientID :
            print(f"Mismatch at mapping {i}: {id1} != {id2}")
        else : match += 1
print(f"Total amount of matches : {match}")
```
