```python
from pydicom import dcmread
from pathlib import Path
import os
import pickle
```

```python
# Load in mapped directory
pkl = Path("/hpcstor6/scratch01/a/anya.tongprasith001/U54REC/release_to_dicom_mapping.pkl")
mapping = pickle.load(open(pkl, "rb"))
```

```python
# Map ID with its dicom file full header
header_map = {}
dir = sorted(mapping.items())[:1000]
for ID, path in dir :
    ds = dcmread(path)
    header_map[ID] = ds
```

```python
# Save dict to pickle file
with open("release_to_header_mapping.pkl","wb") as f:
    pickle.dump(header_map, f)
```
