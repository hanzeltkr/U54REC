```python
from pydicom import dcmread
from pathlib import Path
import os
import pickle
import gzip
```

```python
# Load in mapped directory
pkl = Path("/hpcstor6/scratch01/a/anya.tongprasith001/U54REC/release_to_dicom_mapping_3d.pkl")
mapping = pickle.load(open(pkl, "rb"))
```

```python
# Map ID with its dicom file full header
header_map = {}
files = sorted(mapping.items())
for ID, path in files :
    ds = dcmread(path, stop_before_pixels=True)
    header_map[ID] = ds
```

```python
# Save dict to pickle file in gz
with gzip.open("release_to_header_mapping_3d.pkl.gz", "wb") as f:
    pickle.dump(header_map, f, protocol=pickle.HIGHEST_PROTOCOL)
```
